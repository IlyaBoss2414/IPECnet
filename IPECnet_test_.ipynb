```python
def count_parameters(model):
    """Return the total number of trainable parameters in the model."""
    return sum(p.numel() for p in model.parameters() if p.requires_grad)

def measure_inference_time(model, input_data, n_runs=100):
    """Measure the average inference time over n_runs."""
    model.eval()
    with torch.no_grad():
        # Warm-up
        _ = model(*input_data)
        start_time = time.time()
        for _ in range(n_runs):
            _ = model(*input_data)
        end_time = time.time()
    avg_time = (end_time - start_time) / n_runs
    return avg_time

def measure_memory_usage(model):
    """Return the memory usage of the model in MB (this is a rough estimate)."""
    # Using psutil to get process memory info (in bytes)
    process = psutil.Process()
    mem_bytes = process.memory_info().rss
    mem_mb = mem_bytes / (1024 ** 2)
    return mem_mb
```

```python
from sklearn.metrics import roc_auc_score
from sklearn.metrics import RocCurveDisplay
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
from sklearn.metrics import PrecisionRecallDisplay
from sklearn.metrics import f1_score, recall_score, precision_score, accuracy_score
from sklearn.metrics import precision_recall_curve, f1_score, roc_curve
from sklearn.metrics import cohen_kappa_score



def analize(preds, targets):
    RocCurveDisplay.from_predictions(targets, preds)
    plt.show()

    y_test = targets
    optimal_threshold_roc = 0.5
    # Threshold for converting probabilities to binary predictions
    threshold = optimal_threshold_roc
    y_pred = [1 if prob >= threshold else 0 for prob in preds]


    # Precision-Recall curve
    display = PrecisionRecallDisplay.from_predictions(y_test, preds, name="Model")
    _ = display.ax_.set_title("Precision-Recall Curve")



    df = pd.DataFrame()
    df['score'] = preds
    df['target'] = targets
    df['auc'] = roc_auc_score(targets, preds)
    df['threshold'] = optimal_threshold_roc
    df['score_norm'] = y_pred

    return df


def bootstrap_metric(x, y, metric_fn, samples_cnt=1000, alpha=0.05, random_state=42):
    size = len(x)
    np.random.seed(random_state)
    b_metric = np.zeros(samples_cnt)
    for it in range(samples_cnt):
        poses = np.random.choice(x.shape[0], size=x.shape[0], replace=True)

        x_boot = x[poses]
        y_boot = y[poses]

        m_val = metric_fn(x_boot, y_boot)
        b_metric[it] = m_val

    return b_metric


def compare_bootstrap_metrics_AUC(dfs, metric_fn=lambda x, y: roc_auc_score(y_true=x, y_score=y),
                              n_bootstrap=1000, alpha=0.10):
    """
    Computes bootstrap metrics for multiple models using their provided DataFrames.
    Each DataFrame must contain columns: 'target', 'score', and 'name'.

    Parameters:
      dfs: List of DataFrames, each containing 'target', 'score', and 'name'.
      metric_fn: Function to compute the evaluation metric (default: ROC-AUC).
      n_bootstrap: Number of bootstrap samples.
      alpha: Significance level for confidence intervals (default: 0.10 for 90% CI).

    Returns:
      A dictionary containing bootstrap results for each model.
    """
    results = {}
    ci_results = {}
    scores_combined = []
    labels = []

    n_bootstrap = 1000
    alpha=0.10

    for df in dfs:
        model_name = df['name'].iloc[0]  # Extract model name from DataFrame
        boot_scores = bootstrap_metric(df['target'], df['score'], metric_fn=metric_fn, samples_cnt=n_bootstrap)
        results[model_name] = boot_scores
        ci_results[model_name] = np.quantile(boot_scores, q=[alpha/2, 1 - alpha/2])

        scores_combined.extend(boot_scores)
        labels.extend([model_name] * len(boot_scores))

    # Print confidence intervals
    for model, ci in ci_results.items():
        print(f"{model} ROC-AUC 90%-CI: {ci}")

    # Plot boxplot of the bootstrap results
    plt.figure(figsize=(12, 6))
    sns.boxplot(x=labels, y=scores_combined)
    plt.ylabel("ROC-AUC Score", size=14)
    plt.xticks(rotation=30, ha='right')
    plt.title("Bootstrap ROC-AUC Scores with 90% Confidence Intervals", size=16)
    plt.show()

    return results
```

```python
import time
import torch

def measure_inference_time_dataloader(model, dataloader, n_runs=1):
    """
    Measures the average inference time over the entire DataLoader.

    Parameters:
      model: The PyTorch model.
      dataloader: The DataLoader to iterate over.
      n_runs: Number of complete passes over the DataLoader for averaging.

    Returns:
      Average inference time per full pass over the DataLoader (in seconds).
    """
    model.eval()
    total_time = 0.0

    # Warm-up: run one pass to ensure any lazy initializations are done.
    with torch.no_grad():
        for batch in dataloader:
            # batch is assumed to be a tuple:
            # (inp_PA, inp_PC, PA_features, PC_features, Common_features, target)
            _ = model(*batch[:-1])  # Exclude target
            break  # Only one warm-up batch

    # Now perform n_runs full passes over the DataLoader.
    with torch.no_grad():
        for _ in range(n_runs):
            start_time = time.time()
            for batch in dataloader:
                _ = model(*batch[:-1])
            end_time = time.time()
            total_time += (end_time - start_time)

    avg_time = total_time / n_runs
    return avg_time
```

```python
def extract_scalars_from_events(event_file, tag):
    """
    Extract scalar values for a given tag from a TensorBoard events file.

    Parameters:
        event_file (str): Path to the events file.
        tag (str): Tag name to extract (e.g., 'train_loss' or 'val_loss').

    Returns:
        pd.DataFrame: A DataFrame with columns ['Step', 'Value', 'WallTime'].
    """
    from tensorboard.backend.event_processing import event_accumulator
    ea = event_accumulator.EventAccumulator(event_file)
    ea.Reload()

    if tag not in ea.Tags()['scalars']:
        raise ValueError(f"Tag '{tag}' not found. Available tags: {ea.Tags()['scalars']}")

    scalar_events = ea.Scalars(tag)
    df = pd.DataFrame(scalar_events)
    df = df.rename(columns={'step': 'Step', 'value': 'Value', 'wall_time': 'WallTime'})
    return df

output = "/content/events.out.tfevents.1740415724.s-5f5a47af-2047-44b2-9ff7-e19bb88a0a98.2278.2"

# Example: Extract and plot train_loss and val_loss
df_train_loss = extract_scalars_from_events(output, tag='train_loss')
df_val_loss   = extract_scalars_from_events(output, tag='val_loss')

# Smooth the loss curves using a rolling mean
window_size = 100  # Adjust window size as needed
df_train_loss['Smoothed'] = df_train_loss['Value'].rolling(window=window_size, min_periods=1).mean()
df_val_loss['Smoothed']   = df_val_loss['Value'].rolling(window=window_size, min_periods=1).mean()

plt.figure(figsize=(10,6))
plt.plot(df_train_loss['Step'], df_train_loss['Smoothed'], label='Train Loss', marker='o', linewidth=2)
plt.plot(df_val_loss['Step'], df_val_loss['Smoothed'], label='Validation Loss', marker='o', linewidth=2)
plt.xlabel('Step', fontsize=14)
plt.ylabel('Loss', fontsize=14)
plt.title('IPECnet without chemistry', fontsize=16)
plt.legend(fontsize=12)
plt.grid(True)
plt.tight_layout()
plt.show()
```
